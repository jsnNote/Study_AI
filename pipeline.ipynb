# 파이프라인

```python
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler, 
    MaxAbsScaler,
    Normalizer
)
```

```python
from sklearn.datasets import load_iris
import numpy as np
from pandas import DataFrame
```

```python
iris = load_iris()
iris
```

```python
df = DataFrame(iris["data"], columns=['sepal length', 'sepal width', 'petal length', 'petal width'])
df
```

```python
df['target'] = iris["target"]
df
```

```python
df['class_name'] = df['target'].apply(lambda idx : iris['target_names'][idx])
df
```

```python
x = iris["data"]
y = iris["target"]
```

```python
std_scaler = StandardScaler()
std_scaler.fit(x)
x_scaled = std_scaler.transform(x)
DataFrame(x_scaled, columns=['sepal length', 'sepal width', 'petal length', 'petal width'])
```

```python
minmax_scaler = MinMaxScaler()
print(np.max(x[:, 0]), np.min(x[:, 0]))
minmax_scaler.fit(x)
x_scaled = minmax_scaler.transform(x)
print(DataFrame(x, columns=['sepal length', 'sepal width', 'petal length', 'petal width']))
print(DataFrame(x_scaled, columns=['sepal length', 'sepal width', 'petal length', 'petal width']))
```

```python
maxabs_scaler = MaxAbsScaler()
print(np.max(x[:, 0]), np.min(x[:, 0]))
maxabs_scaler.fit(x)
x_scaled = maxabs_scaler.transform(x)
print(DataFrame(x, columns=['sepal length', 'sepal width', 'petal length', 'petal width']))
print(DataFrame(x_scaled, columns=['sepal length', 'sepal width', 'petal length', 'petal width']))
```

### 스케일링 순서
  1. 훈련용, 테스트용 데이터로 나눈다.
  2. 훈련용 데이터의 통계량을 기준(fit으로 통계량을 저장)
  3. 나누었던 데이터들은 훈련용 데이터의 통계량을 기준으로 스케일링한다(테스트 데이터의 정보를 모델이 알면 과적합하게 훈련하기 때문에 훈련용 데이트를 기준으로 함).
  * 스케일은 단위가 다른 데이터들이 훈련을 했을 때 값이 큰 특성을 중요하게 생각하는 불균형 해소와 단위가 다른 경우 손실함수의 모양이 타원형으로 찌그러져 효율성이 떨어지는 문제를 해소

파이프라인

```python
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.pipeline import Pipeline, make_pipeline
```

```python
## 변수 선택
feat_sel = SelectKBest(f_classif, k=2) # k는 특성 개수
X_selected = feat_sel.fit_transform(x, y)
print('선택된 변수 :', feat_sel.get_feature_names_out())
print(DataFrame(x))
print(DataFrame(X_selected))
```

```python
# 표준화
scaler = StandardScaler()
scaler.fit(X_selected)
X_transformed = scaler.transform(X_selected)
print(DataFrame(X_transformed))
```

### 파이프라인 사용
- 파이프라인을 통해서 변수선택, 표준화, 모델학습을 한 번에 실행할 수 있음.

```python
pipeline = Pipeline([
    ("Feature_Selection", SelectKBest(f_classif, k=2)),
    ("Standardization", StandardScaler())
    ])
pipeline
```

```python
fit_pipeline = pipeline.fit(x, y)
fit_pipeline
```

```python
pipeline.named_steps["Feature_Selection"].get_feature_names_out()
```

```python
makePipeline = make_pipeline(
    SelectKBest(f_classif, k=2),
    StandardScaler()
    )
makePipeline # 작업명을 자동 생성
```
