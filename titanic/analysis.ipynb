# 타이타닉 데이터 분석

```python
import numpy as np
from pandas import DataFrame
import pandas as pd
from matplotlib import pyplot as plt
```

```python
src = "../datasets/titanic/"
train = pd.read_csv(src + "train.csv")
train
```

```python
for index in train.columns:
    print(type(train[index][0]))
```

```python
train.groupby(['Pclass', 'Survived']).size().unstack().plot(kind='bar', color=['salmon', 'lightblue'], width=0.8)
plt.title('타이타닉 승객 생존 여부와 객실 등급(Pclass) 관계')
plt.xlabel("객실 등급")
plt.ylabel('승객 수')
plt.xticks(ticks=[0, 1, 2], labels=['Class 1', 'Class 2', 'Class 3'], rotation=0)
plt.legend(["death", "survival"])
plt.grid(axis='y')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm

# 나이가 결측값인 경우 제거
train = train.dropna(subset=['Age'])

# 나이에 따른 생존 여부 시각화
plt.figure(figsize=(12, 6))
sns.histplot(data=train, x='Age', hue='Survived', multiple='stack', bins=40, palette={0: 'salmon', 1: 'lightblue'}, kde = True, alpha=0.7)

plt.title('타이타닉 승객 생존 여부와 나이(Age) 관계')
plt.xlabel('나이')
plt.ylabel('승객 수')
plt.legend(['생존', '사망'], title='생존 여부')
plt.grid(axis='y')
plt.show()
```

```python
index = ["Pclass", "Sex", "Age", "Survived"]
train[index].sort_values("Survived")
```

```python
#male - 0, female - 1
train["Sex"] = train["Sex"].replace("male", 0)
train["Sex"] = train["Sex"].replace("female", 1)

x = train[["Pclass", "Sex", "Age"]]
y = train["Survived"]
```

```python
print(train.columns)
t = train.dropna(axis=1)
print(t.columns)
t = t.select_dtypes(exclude=['str'])
t_x = t.drop("Survived", axis=1)
t_y = t["Survived"]
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

feat_sel = SelectKBest(f_classif, k=3)
select = feat_sel.fit_transform(t_x, t_y)
print('선택된 변수 :', feat_sel.get_feature_names_out())
```
